# Text Proofreading with MediaPipe Tasks

This notebook will show you how to use the MediaPipe Tasks Python API to proofread input text. The `TextProofreader` task corrects grammar, spelling, and style errors in the text and provides both the proofread text and a list of corrections.

## Running inference

To run inference using the MediaPipe Text Proofreader task, you just need to initialize the `TextProofreader` using the model that you downloaded earlier, and then use that `TextProofreader` to proofread text. You can edit the input text that will be proofread on the side of this section where you see `input_text`.

```python
import os
os.environ["GLOG_minloglevel"] = "2"
os.environ["GOOGLE_STDERRTHRESHOLD"] = "2"

from mediapipe.tasks import python
from mediapipe.tasks.python import text

# Create your base options with the model that was downloaded earlier
base_options = python.BaseOptions(model_asset_path='proofread_quant_200m.litertlm')

# Create the final set of options for the Proofreader
options = text.TextProofreaderOptions(base_options=base_options)

with text.TextProofreader.create_from_options(options) as proofreader:
  # Retrieve the text that will be proofread
  input_text = "I am a studnt." #@param {type:"string"}

  # Run proofreading on the text
  proofreader_result = proofreader.proofread(input_text)

  # Print proofread text
  print(proofreader_result.proofread_text)
```

## Getting list of corrections

Along with the corrected text (`proofread_text`), the `TextProofreaderResult` also includes a list of `corrections`. Each `Correction` object indicates what segment of the text remained the same (`SAME`), was inserted (`INSERTION`), or was deleted (`DELETION`). You can iterate over `proofreader_result.corrections` to inspect the exact edits made by the proofreader:

```python
with text.TextProofreader.create_from_options(options) as proofreader:
  # Retrieve the text that will be proofread
  input_text = "i hope you doing . theyre a great addition on any families." #@param {type:"string"}

  # Run proofreading on the text
  proofreader_result = proofreader.proofread(input_text)

  # Print each correction segment and its type
  for correction in proofreader_result.corrections:
    print(f"[{correction.type.name}]: {correction.text}")
```

# Text Summarizing with MediaPipe Tasks

This notebook will show you how to use the MediaPipe Tasks Python API to summarize input text. The task summarizes long text according to a specified mode such as giving key points (`KEYPOINTS`) or generating a short summary (`TLDR`).

##Preparation

Let's start with installing MediaPipe.

```python
!pip install -q mediapipe
```

After installing your dependencies, you can download the text summarizer model that will be used for this example.

```python
#@title Start downloading here.
!wget -O summarization_quant_200m_2modes.litertlm -q https://storage.googleapis.com/mediapipe-models/text_summarizer/gemma_200m/latest/summarization_quant_200m_2modes.litertlm
```

## Running inference

To run inference using the MediaPipe Text Summarizer task, you just need to initialize the `TextSummarizer` using the model that you downloaded earlier, and then use that `TextSummarizer` to summarize input text. You can edit the text that will be summarized and select the summarization mode (`KEYPOINTS` or `TLDR`) in the options below.

```python
from mediapipe.tasks import python
from mediapipe.tasks.python import text

# Create your base options with the model that was downloaded earlier
base_options = python.BaseOptions(model_asset_path='summarization_quant_200m_2modes.litertlm')

# Set your value for the summarization mode
mode = text.TextSummarizerMode.KEYPOINTS #@param ["text.TextSummarizerMode.KEYPOINTS", "text.TextSummarizerMode.TLDR"] {type:"raw"}

# Create the final set of options for the Summarizer
options = text.TextSummarizerOptions(
    base_options=base_options, mode=mode)

with text.TextSummarizer.create_from_options(options) as summarizer:
  # Retrieve the text that will be summarized
  input_text = "This station is in the centre of the European quarter of Brussels, being adjacent to the Berlaymont building (headquarters of the European Commission), the Justus Lipsius building (Council of the European Union) and numerous other EU offices. The station is named after the area around the Schuman roundabout, which was named after Robert Schuman. The station sits on the Rue de la Loi/Wetstraat, a major city thoroughfare, and is close to Parc du Cinquantenaire/Jubelpark." #@param {type:"string"}

  # Run summarization on the text
  summarizer_result = summarizer.summarize(input_text)

  # Print the summary
  print(summarizer_result.summary)
```
